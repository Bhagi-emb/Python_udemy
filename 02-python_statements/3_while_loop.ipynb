# **WHILE LOOP**

```python
x = 0
while(x < 4) :
    #print("x value : ",x)
    print(f'x value {x}')
    x += 1
else :
    print(" x is greater tha 4")

```

# **Break continue pass**

```python
#PASS -  Does noting pass it is used to avoid syntax error
x =[1,2,3]
for _ in x:
    pass   # without this pass it cause syntax error
print("end of lines")
```

```python
#continue - goes to top of the closest enclosing loop
str ='hello'
for letter in str:
    if letter == 'h':
        continue
    print(f'str : {letter}')
```

```python
# break - breaks out of the closest enclosing loop
for letter in str:
    if letter =='o':
        break
    print(f'str : {letter}')
```
